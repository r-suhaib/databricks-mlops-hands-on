```python
## Imports

from src.monitoring.feature_monitor import (
    calculate_feature_statistics
)

from src.monitoring.drift_detector import (
    detect_mean_drift
)
```

```python
## Load data


SOURCE_TABLE = (
    "tep_anomaly.served.training_base"
)

df = spark.table(
    SOURCE_TABLE
)
```

```python
## Select monitored features

monitored_features = [
    "xmeas_1",
    "xmeas_2",
    "xmeas_3",
    "xmeas_1_delta1",
    "xmeas_2_delta1",
    "xmeas_3_delta1"
]
```

```python
## generate baseline stats

baseline_stats = (
    calculate_feature_statistics(
        df,
        monitored_features
    )
)
```

```python
## generate current stats

current_stats = (
    calculate_feature_statistics(
        df,
        monitored_features
    )
)
```

```python
## detect drift

drift_results = (
    detect_mean_drift(
        baseline_stats,
        current_stats,
        threshold=20
    )
)

drift_results
```

```python
## print drift_results

for feature, result in drift_results.items():

    print(
        f"{feature} | "
        f"Drift %: "
        f"{result['drift_percent']:.2f} | "
        f"Detected: "
        f"{result['drift_detected']}"
    )
```

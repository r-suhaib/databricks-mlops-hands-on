```python
## imports

from src.monitoring.feature_monitor import (
    calculate_feature_statistics
)

from src.monitoring.monitoring_writer import (
    write_feature_statistics
)
```

```python
## Load dataset

SOURCE_TABLE = (
    "tep_anomaly.served.training_base"
)

df = spark.table(
    SOURCE_TABLE
)
```

```python
## Define monitored features

monitored_features = [
    "xmeas_1",
    "xmeas_2",
    "xmeas_3",
    "xmeas_1_delta1",
    "xmeas_2_delta1",
    "xmeas_3_delta1"
]
```

```python
## calculate calculate statistics 

statistics = (
    calculate_feature_statistics(
        df,
        monitored_features
    )
)

statistics
```

```python
## Persist statistics

MONITORING_TABLE = (
    "tep_anomaly.served.feature_monitoring_history"
)

write_feature_statistics(
    spark=spark,
    statistics=statistics,
    table_name=MONITORING_TABLE
)
```

```python
# validate results 

display(
    spark.sql(
        """
        SELECT *
        FROM tep_anomaly.served.feature_monitoring_history
        ORDER BY monitoring_timestamp DESC
        """
    )
)
```

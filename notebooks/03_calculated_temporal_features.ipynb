```python
from src.feature_engineering.temporal_features import (
    create_temporal_features
)
```

```python
df = spark.table(
    "tep_anomaly.source.faulty_training"
)
```

```python
features_df = create_temporal_features(df)
```

```python
(
    features_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "tep_anomaly.calculated.sensor_temporal_features"
    )
)
```

```python
row_count = features_df.count()
print(
        f"Calculated dataset rows: {row_count:,}"
        )
```

```python
assert row_count > 0
```

```python
# check duplicate keys
from pyspark.sql import functions as F

duplicates = (
    features_df
    .groupBy(
        "simulationRun",
        "sample"
    )
    .count()
    .filter(F.col("count") > 1)
    .count()
)

print(
    f"Duplicate keys: {duplicates}"
)
```

```python
assert duplicates == 0
```

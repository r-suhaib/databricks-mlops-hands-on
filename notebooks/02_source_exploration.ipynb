```python
faultfree_training_df = spark.table(
    "tep_anomaly.source.faultfree_training"
)

faultfree_testing_df = spark.table(
    "tep_anomaly.source.faultfree_testing"
)

faulty_training_df = spark.table(
    "tep_anomaly.source.faulty_training"
)

faulty_testing_df = spark.table(
    "tep_anomaly.source.faulty_testing"
)
```

```python
datasets = {
    "faultfree_training": faultfree_training_df,
    "faultfree_testing": faultfree_testing_df,
    "faulty_training": faulty_training_df,
    "faulty_testing": faulty_testing_df,
}

for name, df in datasets.items():

    print("=" * 60)
    print(f"Dataset: {name}")
    print(f"Rows    : {df.count():,}")
    print(f"Columns : {len(df.columns)}")
```

```python
metadata_columns = [
    "faultNumber",
    "simulationRun",
    "sample"
]

for name, df in datasets.items():

    print("=" * 60)
    print(name)

    df.select(metadata_columns).show(5)
```

```python
display(
    faultfree_training_df.groupBy(
        "faultNumber"
    ).count()
)
```

```python
display(
    faulty_training_df.groupBy(
        "faultNumber"
    ).count()
    .orderBy("faultNumber")
)
```

```python
for name, df in datasets.items():

    simulation_runs = (
        df
        .select("simulationRun")
        .distinct()
        .count()
    )

    print(
        f"{name}: {simulation_runs:,} simulation runs"
    )
```

```python
display(
    faultfree_training_df
    .filter("simulationRun = 1")
    .orderBy("sample")
    .limit(100)
)
```

```python
xmeas_columns = [
    c for c in faultfree_training_df.columns
    if c.startswith("xmeas_")
]

xmv_columns = [
    c for c in faultfree_training_df.columns
    if c.startswith("xmv_")
]

print(
    f"Measured Variables: {len(xmeas_columns)}"
)

print(
    f"Manipulated Variables: {len(xmv_columns)}"
)

print("\nMeasured Variables")
print(xmeas_columns)

print("\nManipulated Variables")
print(xmv_columns)
```

```python
display(
    faultfree_training_df.select(
        "xmeas_1",
        "xmeas_2",
        "xmeas_3",
        "xmv_1",
        "xmv_2"
    ).summary()
)
```

```python
from pyspark.sql.functions import col, count, when

null_counts = faultfree_training_df.select(
    [
        count(
            when(col(c).isNull(), c)
        ).alias(c)
        for c in faultfree_training_df.columns
    ]
)

display(null_counts)
```

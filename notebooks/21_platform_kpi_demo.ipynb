```python
from src.platform.platform_kpis import (
    build_kpi_records
)

total_models = (
    spark.table(
        "tep_anomaly.served.model_inventory"
    )
    .count()
)

active_models = (
    spark.table(
        "tep_anomaly.served.model_inventory"
    )
    .filter("active_flag = true")
    .count()
)

monitoring_alerts = (
    spark.table(
        "tep_anomaly.served.monitoring_alerts"
    )
    .count()
)

records = build_kpi_records(
    total_models,
    active_models,
    monitoring_alerts
)

kpi_df = spark.createDataFrame(
    records
)

(
    kpi_df.write
    .mode("append")
    .saveAsTable(
        "tep_anomaly.served.platform_kpis"
    )
)

display(
    spark.table(
        "tep_anomaly.served.platform_kpis"
    )
)
```

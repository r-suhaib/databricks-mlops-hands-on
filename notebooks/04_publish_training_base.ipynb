```python
# read calculated delta table
calculated_df = spark.table(
    "tep_anomaly.calculated.sensor_temporal_features"
)
```

```python
# build served dataset
from src.served.training_base import (
    create_training_base
)

served_df = create_training_base(
    calculated_df
)
```

```python
# write served table
(
    served_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "tep_anomaly.served.training_base"
    )
)
```

```python
# add served layer validation

def validate_training_base(df):

    required_columns = [
        "faultNumber",
        "simulationRun",
        "sample"
    ]

    missing = [
        c for c in required_columns
        if c not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns: {missing}"
        )
```

```python
# run validation
validate_training_base(
    served_df
)
```
